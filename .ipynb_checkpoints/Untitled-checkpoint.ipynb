```python
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
# create objects
fig = plt.figure(constrained_layout=True)
gs = GridSpec(2, 4, figure=fig)
 
# create sub plots as grid
ax2 = fig.add_subplot(gs[0, :-2])
ax3 = fig.add_subplot(gs[0:, -2:])
ax4 = fig.add_subplot(gs[-1, 0])
ax5 = fig.add_subplot(gs[-1, -3])
 
# depict illustration
fig.suptitle("GridSpec")
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
# create objects
fig = plt.figure(constrained_layout=True)
gs = GridSpec(2, 3, figure=fig)
gs.set_width_ratios([1,1,2])
 
# create sub plots as grid
ax2 = fig.add_subplot(gs[0, :-1])
ax3 = fig.add_subplot(gs[0:, -1:])
ax4 = fig.add_subplot(gs[-1, 0])
ax5 = fig.add_subplot(gs[-1, -2])
 
# depict illustration
fig.suptitle("GridSpec")
plt.show()


```

```python
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
# create objects
fig = plt.figure(constrained_layout=True)
gs = GridSpec(2, 4, figure=fig)
#gs.set_width_ratios([1,1,1,1])
 
# create sub plots as grid
ax1 = fig.add_subplot(gs[0:, :-2])
ax2 = fig.add_subplot(gs[0, -2:])
ax3 = fig.add_subplot(gs[1, -2:])
 
# depict illustration
fig.suptitle("GridSpec")
plt.show()
```

ax1 = fig.add_subplot(gs[0, :])       # First row, spans all columns
ax2 = fig.add_subplot(gs[1, :-1])     # Second row, spans first two columns
ax3 = fig.add_subplot(gs[1:, -1])     # Second and third rows, last column
ax4 = fig.add_subplot(gs[-1, 0])      # Third row, first column
ax5 = fig.add_subplot(gs[-1, -2])     # Third row, second column
